````python
import json
import re

def extract_ts_texts_with_delimiter(input_jsonl_path: str, output_txt_path: str, column_name: str = "content", delimiter: str = "\n" + "-"*80 + "\n") -> list:
    """
    Reads a JSONL file where each line is a JSON object containing at least a 'content' key.
    Extracts the 'content' values, converts literal '\\n' to real newlines,
    separates each snippet by the given delimiter, writes them to a text file,
    and returns the list of snippets.
    """
    contents = []
    with open(input_jsonl_path, 'r') as infile:
        for line in infile:
            try:
                rec = json.loads(line)
                content = rec.get(column_name)
                if content:
                    # Convert literal '\n' to actual newlines
                    content = content.replace('\\n', '\n')
                    contents.append(content)
            except json.JSONDecodeError:
                continue

    with open(output_txt_path, 'w') as outfile:
        for i, snippet in enumerate(contents):
            outfile.write(snippet)
            if i < len(contents) - 1:
                outfile.write(delimiter)

    return contents

def jsonl_to_pipe_txt(
    input_path: str,
    output_path: str,
    fields: list[str],
    delimiter: str = "|",
    write_header: bool = True
):
    with open(input_path, "r", encoding="utf-8") as in_f, \
         open(output_path, "w", encoding="utf-8") as out_f:

        if write_header:
            out_f.write(delimiter.join(fields) + "\n")

        for line in in_f:
            rec = json.loads(line)
            vals = []
            for key in fields:
                v = rec.get(key, "")
                if isinstance(v, list):
                    v = ",".join(map(str, v))
                else:
                    v = str(v)
                v = v.replace("\n", "\\n")
                vals.append(v)

            text = delimiter.join(vals)
            raw = json.dumps(text, ensure_ascii=False)
            cleaned_text = raw.replace(r'\n', '\n')
            out_f.write(cleaned_text + "\n")

    print(f"Wrote {len(fields)}‑column {delimiter!r}-delimited file → {output_path}")

def clean_response(resp: str) -> str:
    """
    Remove markdown fences (```ts``` / ```), replace • bullets, and normalize blank lines.
    """
    lines = []
    for line in resp.splitlines():
        if line.strip().startswith("```"):
            continue
        line = line.replace("•", "-").rstrip()
        lines.append(line)
    out = "\n".join(lines)
    out = re.sub(r"\n{3,}", "\n\n", out)
    return out.strip()

def convert_jsonl_to_txt(input_path: str, output_path: str):
    """
    Reads a JSONL with keys [seed, description, concepts, instruction, response]
    and writes a plain‑text file.
    """
    with open(input_path, "r", encoding="utf-8") as fin, \
         open(output_path, "w", encoding="utf-8") as fout:
        for lineno, line in enumerate(fin, 1):
            try:
                rec = json.loads(line)
            except json.JSONDecodeError as e:
                print(f" Skipping invalid JSON on line {lineno}: {e}", file=sys.stderr)
                continue

            seed        = rec.get("seed", "").strip()
            desc        = rec.get("description", "").strip()
            concepts    = rec.get("concepts", [])
            instr       = rec.get("instruction", "").strip()
            response    = rec.get("response", "")

            fout.write("seed:\n")
            fout.write(f"{seed}\n\n")

            fout.write("description: ")
            fout.write(f"{desc}\n\n")

            fout.write("concepts: ")
            fout.write(", ".join(concepts) + "\n\n")

            fout.write("instruction: ")
            fout.write(f"{instr}\n\n")

            fout.write("response:\n")
            fout.write(clean_response(response) + "\n")

            fout.write("\n" + "="*80 + "\n\n")

    print(f" Converted → {output_path}")

def clean_concepts(raw_concepts: list[str]) -> list[str]:
    """
    Cleans a list of raw concepts:
    - strips whitespace
    - removes numeric-only or too-short entries
    - deduplicates while preserving order
    """
    cleaned = []
    seen = set()
    for tok in raw_concepts:
        c = tok.strip()
        if re.fullmatch(r'\d+', c): continue
        if len(c) < 2: continue
        if c not in seen:
            cleaned.append(c)
            seen.add(c)
    return cleaned
````

````python
import numpy
import pandas
from vllm import LLM, SamplingParams
import datasets

def parse_args_generate():
    p = argparse.ArgumentParser("Step 2 S→C→I→R with Starcoder2")
    p.add_argument("--model", default="bigcode/starcoder2-3b", help="vLLM name of HuggingFace model")
    p.add_argument("--seed", type=int, default=42)
    p.add_argument("--rate-sleep", type=float, default=0, help="helps between calls to avoid throttling")
    p.add_argument("--batch_size", type=int, default=32, help="Number of prompts per vLLM call")
    args = p.parse_args([])  # replace [] with sys.argv[1:] in real use

    return args

def init_llm_generate(model_name):
    return LLM(
        model_name,
        gpu_memory_utilization=0.95,
        enforce_eager=True
    )
````

````python
def llm_generate(model, prompt, max_tokens=512, stop="###", temp=0.0):
    """
    Generate a response using the provided LLM and prompt.

    Args:
        model: An instance of vLLM's LLM class.
        prompt (str): The prompt to feed into the model.
        max_tokens (int): Maximum number of tokens to generate.
        stop (str or list): Stop sequence(s) to end generation.
        temp (float): Temperature for randomness.

    Returns:
        str: The generated text output.
    """
    sampling_params = SamplingParams(
        temperature=temp,
        max_tokens=max_tokens,
        stop=stop if isinstance(stop, list) else [stop],
        top_p=0.95,  # Optional: helps avoid repetitive completions
        n=1
    )

    outputs = model.generate(prompt, sampling_params)

    # Safety: Handle if output or text is unexpectedly missing
    try:
        text = outputs[0].outputs[0].text.strip()
    except Exception as e:
        print(f" Failed to extract text: {e}")
        return ""

    print(f"Response:\n{text}")
    return text
````

````python
import re

def prompt_s_to_c(ts_code: str) -> str:
    return f"""
You are an expert TypeScript concept extraction assistant.
Ignore any leading comments in the snippet. Your task is to identify the TypeScript concepts used in the code, such as function declaration, arrow functions, async/await, interface, class, type annotations, destructuring, promises, loops, error handling, imports, exports, DOM manipulation, etc.
Return ONLY a comma-separated list of these concepts, without any additional explanation or labels.

Example 1:
Input:
```ts
// Fetch user data from an API
async function getUser(id: number): Promise<User> {{
  const res = await fetch(`/api/users/${{id}}`);
  return await res.json();
}}
```
Output:
async function, type annotation, await, fetch, Promise
###

Example 2:
Input:
```ts
// Add two numbers using an arrow function
const add = (a: number, b: number): number => a + b;
```
Output:
arrow function, type annotation, const
###

Now process this TypeScript:
```ts
{ts_code}
```
Output:
"""

def prompt_c_to_i(ts_code, concepts):
    return f"""You are an expert TypeScript instruction writer.
Write a single, clear natural-language instruction asking someone to write a TypeScript function or script that matches the description and uses exactly those concepts.
Return ONLY the instruction, without any additional label or TypeScript code.

Example 1:
TypeScript:
```ts
// Get the names from an array of users
const names = users.map(user => user.name);
```
Concepts: ["array", "map", "arrow function", "const"]
Instruction: Write a TypeScript function that extracts the names from an array of user objects using array methods and arrow functions.
###

Now process below TypeScript, Description, Concepts and then Return the Instruction:
TypeScript:
```ts
{ts_code}
```
Concepts: {concepts}
Instruction:
"""

def prompt_i_to_r(instruction: str) -> str:
    return f"""
You are an expert TypeScript assistant.
Given a TypeScript-related instruction, generate a response with four sections:
Reasoning: Explain your step-by-step thought process.
Solution: Provide the exact TypeScript code.
Explanation: Describe why this solution works.
Tests: Include a short test or usage example to verify the code.

Example 1
Input:
Write a TypeScript function that checks if a number is even.

Output:
-- Reasoning:
We need to check whether a number is divisible by 2. This can be done using the modulo operator.

-- Solution:
```ts
function isEven(n: number): boolean {{
  return n % 2 === 0;
}}
```

-- Explanation:
This function uses `%` to check if the number leaves a remainder when divided by 2. If it's zero, the number is even.

-- Tests:
console.log(isEven(4)); // true
console.log(isEven(7)); // false
###

Example 2
Input:
Create a TypeScript class called Person with a constructor and a greet method.

Output:
-- Reasoning:
We need a class with a constructor to initialize properties and a method to return a greeting string.

-- Solution:
```ts
class Person {{
  constructor(public name: string) {{}}

  greet(): string {{
    return `Hello, my name is ${{this.name}}`;
  }}
}}
```

-- Explanation:
The constructor assigns a public field. The greet method returns a string using template literals.

-- Tests:
const p = new Person("Alice");
console.log(p.greet()); // "Hello, my name is Alice"
###

Now process this Input:
Input:
{instruction}

Output:
"""

def extract_description(ts_code: str) -> str:
    """Extracts block comment description from the top of a TypeScript code snippet."""
    m = re.search(r'/\*(.*?)\*/', ts_code, re.DOTALL)
    return m.group(1).strip() if m else ""
````

````python
import json
import time
import gc
from vllm import SamplingParams
from tqdm import tqdm

IN_FILE  = "/content/Step2_3_Data/seed3.jsonl"
OUT_FILE = "/content/Step2_3_Data/ts_s2c_output.jsonl"

with open(IN_FILE, "r", encoding="utf-8") as f:
    ds = [json.loads(line) for line in f]

# Tokenizer may be used for optional validation or token count logging
tokenizer = llm.get_tokenizer()
out_f = open(OUT_FILE, "w")
N = len(ds)
batch_size = args.batch_size

# Build prompt list
records = []
prompts = []
for rec in ds:
    code = rec["content"].strip()
    records.append(rec)
    prompts.append(prompt_s_to_c(code))

# Batch generation
for i in tqdm(range(0, N, batch_size), desc="S→C batches"):
    batch_recs = records[i : i + batch_size]
    batch_prs  = prompts [i : i + batch_size]

    outs = llm.generate(
        batch_prs,
        SamplingParams(
            temperature=0.0,
            max_tokens=30,
            stop="###"
        ),
    )

    # Collect outputs and write to file
    for rec, out in zip(batch_recs, outs):
        text = out.outputs[0].text.strip()
        concepts = [c.strip() for c in text.split(",") if c.strip()]
        concepts = clean_concepts(concepts)
        rec_out = {
            "seed":        rec["content"],
            "description": extract_description(rec["content"].strip()),
            "concepts":    concepts
        }
        out_f.write(json.dumps(rec_out) + "\n")
        print(f"Response: {concepts}")

    gc.collect()
    time.sleep(0.1)

print("Step S->C is completed →", OUT_FILE)
````
